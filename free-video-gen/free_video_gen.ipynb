## Pick a model
- `ltx` – fastest (≈1–3 min/clip on T4), supports image-to-video
- `wan` – best quality for a free GPU (Wan 2.1 1.3B, ≈5–10 min/clip on T4)
- `wan22` – 720p, needs a bigger GPU (A100/L4 or 24 GB)
- `cogvideox` – lightweight fallback
